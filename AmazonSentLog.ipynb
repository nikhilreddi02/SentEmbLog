```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
import nltk
from nltk.corpus import stopwords
nltk.download('stopwords')
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
df = pd.read_csv('AmazonReview.csv')
```

```python
df
```

```python
df['Sentiment'].value_counts()
```

```python
df['Sentiment'] = df['Sentiment'].apply(lambda x : 1 if x > 3 else 0)
```

```python
stop_words = stopwords.words('english')
```

```python
df['Review'] = df['Review'].astype(str).apply(lambda x: ' '.join([word for word in x.split() if word not in (stop_words)]))
```

```python
df['Review'].head()
```

```python
X = df['Review'].tolist()
X
```

```python
from sentence_transformers import SentenceTransformer
embedding = SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2').encode(X)
```

```python
embedding.shape
```

```python
Y = df['Sentiment'].tolist()
```

```python
Y[0]
```

```python
X_train,x_test,Y_train,y_test = train_test_split(embedding,Y,test_size=0.2,random_state=42,stratify=Y)
```

```python
log = LogisticRegression()
```

```python
log.fit(X_train,Y_train)
```

```python
ypred = log.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
print(accuracy_score(y_test,ypred))
```

```python
sent = ['I dont the product because it was too high but not at all usefull ']
sent_embedding = SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2').encode(sent)
ypred = log.predict(sent_embedding)
print(ypred)
```

```python
sent_embedding.shape
```

```python
from sklearn.ensemble import RandomForestClassifier
rf = RandomForestClassifier()
rf.fit(X_train,Y_train)
```

```python
ypred = rf.predict(x_test)
print(accuracy_score(y_test,ypred))
#
```
